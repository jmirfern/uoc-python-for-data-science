# Ex 1. (0,4p) Càrrega del dataset i anàlisi exploratori de les dades

La descripció de les columnes del dataset es:
<pre>
FTHG - Number of goals scored by Home Team.
FTAG - Number of goals scored by Away Team.
FTR - Full time result.
HTHG - Number of goals scored by Home Team at Half time.
HTAG - Number of goals scored by Away Team at Half time.
HTR - Half time result.
H - Home Team.
A - Away Team.
D - Draw.
</pre>
- Defineix la funció *load_and_eda(file)*, que carrega el dataset i elimina les columnes "HTHG", "HTAG",
"HTR". La funció també mostra els primers i últims valors del dataset, i la informació més rellevant.
- Executa la funció.
- Defineix la funció *plot_home_away_goals(data)*, que conté una figura amb dos plots. En la gràfica es
mostra com és la distribució de gols marcats pels equips de casa i pels equips de fora. Ajuda: pots utilitzar
`plt.boxplot`
- Executa la funció

```python
from datetime import datetime
from pathlib import Path
from typing import Union

import matplotlib.pyplot as plt
import pandas as pd
from IPython.display import display

# Substitueix aquest valor pel teu nom abans de generar les gràfiques finals.
nom_alumne = "Jonathan_Mir"
date_time = datetime.now().strftime("%Y%m%d_%H%M%S")

PROJECT_DIR = Path.cwd()
DATA_FILE = PROJECT_DIR / "data" / "LaLiga_Matches.csv"
if not DATA_FILE.exists():
    PROJECT_DIR = Path("Activity_4")
    DATA_FILE = PROJECT_DIR / "data" / "LaLiga_Matches.csv"

IMG_DIR = PROJECT_DIR / "img"


def load_and_eda(file: Union[str, Path]) -> pd.DataFrame:
    """Carrega el dataset, elimina columnes de descans i mostra una EDA bàsica.

    Args:
        file: Ruta relativa o absoluta del fitxer CSV amb els resultats de LaLiga.

    Returns:
        Dataset carregat sense les columnes HTHG, HTAG i HTR.
    """
    data = pd.read_csv(file)
    data = data.drop(columns=["HTHG", "HTAG", "HTR"], errors="ignore")

    print("Primers valors del dataset:")
    display(data.head())

    print("Últims valors del dataset:")
    display(data.tail())

    print("Informació del dataset:")
    data.info()

    print("Resum estadístic de les variables numèriques:")
    display(data.describe())

    print("Valors nuls per columna:")
    display(data.isnull().sum())

    return data


def plot_home_away_goals(data: pd.DataFrame) -> None:
    """Genera i desa un boxplot dels gols locals i visitants.

    Args:
        data: Dataset de partits amb les columnes FTHG i FTAG.
    """
    IMG_DIR.mkdir(parents=True, exist_ok=True)

    fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(10, 5), sharey=True)

    axes[0].boxplot(data["FTHG"].dropna())
    axes[0].set_title("Gols equip local")
    axes[0].set_ylabel("Nombre de gols")
    axes[0].set_xticks([1])
    axes[0].set_xticklabels(["Local"])

    axes[1].boxplot(data["FTAG"].dropna())
    axes[1].set_title("Gols equip visitant")
    axes[1].set_xticks([1])
    axes[1].set_xticklabels(["Visitant"])

    fig.suptitle("Distribució de gols locals i visitants")
    fig.tight_layout()

    output_file = IMG_DIR / f"grafica_ex1_{nom_alumne}_{date_time}.png"
    plt.savefig(output_file, dpi=300, bbox_inches="tight")
    plt.show()


# Execució de l'exercici 1
data = load_and_eda(DATA_FILE)
plot_home_away_goals(data)
```
