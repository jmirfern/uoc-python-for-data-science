4. Crea una nova columna anomenada **"classificacio_volum"** que categoritzi el volum d'aigua de la següent manera:
   - Si "Volum embassat (hm3)" > 150 hm³ → "Alt"
   - Si 50 ≤ "Volum embassat (hm3)" ≤ 150 hm³ → "Mitjà"
   - Si "Volum embassat (hm3)" < 50 hm³ → "Baix"
5. Mostra la distribució de les classificacions de volum (quants registres hi ha a cada categoria).

<span style="font-family: Courier New; background-color: #82b74b; color: #000000; padding: 3px;">SM</span> **(2 punts)**

```python
# 1. Percentatge mitjà per embassament
volum_mitja = df.groupby('Estació')['Percentatge volum embassat (%)'].mean().sort_values(ascending=False)
print("Percentatge mitjà per embassament:")
print(volum_mitja)
```

```python
# 2. Embassament amb percentatge d'ompliment mitjà més alt i més baix
percentatge_mitja = df.groupby('Estació')['Percentatge volum embassat (%)'].mean()

embassament_mes_ple = percentatge_mitja.idxmax()
valor_mes_ple = percentatge_mitja.max()

embassament_mes_buit = percentatge_mitja.idxmin()
valor_mes_buit = percentatge_mitja.min()

print(f"Embassament amb percentatge d'ompliment mitjà MÉS ALT: {embassament_mes_ple} ({valor_mes_ple:.2f}%)")
print(f"Embassament amb percentatge d'ompliment mitjà MÉS BAIX: {embassament_mes_buit} ({valor_mes_buit:.2f}%)")
```

```python
# 3. Comptar registres per any
registres_per_any = df['Any'].value_counts().sort_index()
print("Nombre de registres per any:")
print(registres_per_any)
```

```python
# 4. Crear columna Classificació_Volum basada en el volum d'aigua
def classificar_volum(volum: float) -> str:
    """Funció per classificar el volum d'aigua en categories."""
    if volum > 150:
        return "Alt"
    elif 50 <= volum <= 150:
        return "Mitjà"
    else:
        return "Baix"

df['classificacio_volum'] = df['Volum embassat (hm3)'].apply(classificar_volum)

print("Primeres files amb la columna classificacio_volum:")
print(df[['Estació', 'Volum embassat (hm3)', 'classificacio_volum']].head(10))
```

```python
# 5. Distribució de les classificacions de volum
distribucio_volum = df['classificacio_volum'].value_counts()
print("Distribució de les classificacions de volum:")
print(distribucio_volum)
```

```python
# 6. Ordenar de manera ascendent i desar en un OrderedDict
from collections import OrderedDict

classificacio_ordenada = OrderedDict(sorted(distribucio_volum.items(), key=lambda x: x[1], reverse=False))

print(classificacio_ordenada)
```

## Exercici 3

Utilitzant el DataFrame obtingut a l'Exercici 1:

1. Crea una **funció anomenada `analitzar_embassaments_per_rang`** que rebi com a paràmetres:
   - El DataFrame
   - Un rang de percentatges d'ompliment: `min_percentatge` i `max_percentatge`
2. La funció ha de retornar un **diccionari** amb:
   - **"recompte"**: nombre total de mesures dins el rang de percentatge
   - **"volum_mitja"**: volum mitjà d'aigua (en hm³) per a aquest rang
   - **"nivell_absolut_mitja"**: nivell absolut mitjà (en msnm) per a aquest rang
   - **"embassaments_implicats"**: llista d'embassaments únics que tenen mesures en aquest rang
3. Prova la funció amb almenys **dos exemples diferents** i mostra els resultats.

**Pista:** Pots utilitzar un diccionari per retornar múltiples valors des de la funció.

**Pista 2:** Utilitza `filter()` o una condició lògica amb `&` per filtrar pel rang de percentatges.

<span style="font-family: Courier New; background-color: #82b74b; color: #000000; padding: 3px;">SM</span> **(1,5 punts)**

```python
def analitzar_embassaments_per_rang(df: pd.DataFrame, min_percentatge: float, max_percentatge: float) -> dict:
    """
    Analitza mesures d'embassaments dins d'un rang de percentatges d'ompliment.
    
    Paràmetres:
    - df: DataFrame amb les dades d'embassaments
    - min_percentatge: percentatge mínim d'ompliment
    - max_percentatge: percentatge màxim d'ompliment
    
    Retorna:
    - Diccionari amb estadístiques del rang
    """
    
    filtre = df[(df['Percentatge volum embassat (%)'] >= min_percentatge) & 
                (df['Percentatge volum embassat (%)'] <= max_percentatge)]
    
    resultat = {
        "recompte": len(filtre),
        "volum_mitja": round(filtre['Volum embassat (hm3)'].mean(), 2),
        "nivell_absolut_mitja": round(filtre['Nivell absolut (msnm)'].mean(), 2),
        "embassaments_implicats": filtre['Estació'].unique().tolist()
    }
    
    return resultat


resultat_alt = analitzar_embassaments_per_rang(df, 80, 100)
print("Anàlisi per embassaments amb ompliment alt (80-100%):")
print(resultat_alt)
resultat_alt = analitzar_embassaments_per_rang(df, 40, 60)
print("Anàlisi per embassaments amb ompliment mitjà (40-60%):")
print(resultat_alt)
```

## Exercici 5

En aquest exercici crearàs classes per modelar un sistema de **gestió de productes en un magatzem**.

1. Crea una **classe `Producte`** amb els següents atributs:
   - `nom` (nom del producte)
   - `preu` (preu unitari)
   - `stock` (quantitat disponible)

   Implementa els següents mètodes:
   - `__init__()`: constructor que inicialitza els atributs
   - `__str__()`: retorna una representació llegible del producte format: "Nom: {nom} | Preu: {preu} | Stock: {stock}"
   - `obtenir_valor_stock()`: retorna el valor total de l'stock (preu × quantitat)

2. Crea una **classe `Magatzem`** que gestioni productes:
   - Atribut `productes` (llista d'objectes Producte)
   - Mètode `afegir_producte(producte: Producte)`: afegeix un producte al magatzem
   - Mètode `cercar_producte(nom: str)`: cerca i retorna un producte pel nom (o None si no existeix)
   - Mètode `obtenir_valor_total_magatzem()`: retorna el valor total de tot l'stock

3. **Crea un magatzem amb productes**, seguint aquests passos:
   - Afegeix almenys 3 productes diferents
   - Cerca un producte pel nom al magatzem
   - Imprimeix el producte i el seu valor en stock
   - Mostra el valor total del magatzem

<span style="font-family: Courier New; background-color: #82b74b; color: #000000; padding: 3px;">SM</span> **(3 punts)**

```python
class Producte:
    """
    Classe que representa un producte en el magatzem.
    Cada producte té nom, preu i quantitat en stock.
    """
    
    def __init__(self, nom: str, preu: float, stock: int) -> None:
        """
        Constructor de la classe Producte.
        
        Paràmetres:
        - nom: nom descriptiu del producte
        - preu: preu unitari en euros
        - stock: quantitat disponible en magatzem
        """
        self.nom = nom
        self.preu = preu
        self.stock = stock
    
    def __str__(self) -> str:
        """Retorna una representació llegible del producte."""
        return f"Nom: {self.nom} | Preu: {self.preu:.2f} | Stock: {self.stock}"
    
    def afegir_stock(self, quantitat: int) -> None:
        """Afegeix quantitat a l'stock del producte."""
        if quantitat > 0:
            self.stock += quantitat
    
    def obtenir_valor_stock(self) -> float:
        """Retorna el valor total de l'stock (preu × quantitat)."""
        return self.preu * self.stock
```

```python
class Magatzem:
    """
    Classe que representa un magatzem que gestiona múltiples productes.
    Permet afegir, cercar i generar informes de productes.
    """
    
    def __init__(self) -> None:
        """Constructor que inicialitza la llista buida de productes."""
        self.productes: list[Producte] = []
    
    def afegir_producte(self, producte: Producte):
        """Afegeix un producte al magatzem."""
        if any(p.nom == producte.nom for p in self.productes):
            print(f"Ja existeix un producte {producte.nom}")
        self.productes.append(producte)
    
    def cercar_producte(self, nom: str) -> Producte | None:
        """
        Cerca i retorna un producte pel nom.
        Retorna None si no es troba.
        """
        for producte in self.productes:
            if producte.nom == nom:
                return producte
        return None
    
    def obtenir_valor_total_magatzem(self) -> float:
        """Retorna el valor total de tot l'stock del magatzem."""
        return sum(producte.obtenir_valor_stock() for producte in self.productes)
```

```python
magatzem = Magatzem()

portatil = Producte(nom="Portàtil 1", preu=899.99, stock=5)
rato = Producte(nom="Ratolí UOC", preu=45.50, stock=20)
teclat = Producte(nom="Teclat PAC1", preu=125.00, stock=8)

magatzem.afegir_producte(portatil)
magatzem.afegir_producte(rato)
magatzem.afegir_producte(teclat)
```

```python
print("\n--- Cerca de Producte ---\n")
producte_cercat = magatzem.cercar_producte("Ratolí UOC")
if producte_cercat:
    print(f"Trobat: {producte_cercat}")
    print(f"Valor de l'stock: {producte_cercat.obtenir_valor_stock():.2f}")
else:
    print("Producte no trobat")

valor_total = magatzem.obtenir_valor_total_magatzem()
print(f"Valor total invertit en stock: {valor_total:.2f}")
```
