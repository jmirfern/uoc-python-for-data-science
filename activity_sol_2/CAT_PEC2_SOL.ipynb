1.2: (0,5 punts) Crea una funció anomenada Calcular_Volumen_Empleo que afegeixi al dataframe una nova columna Total_Jobs_Volume (suma de feines creades i eliminades) i torni el dataframe modificat.

1.3: (0,5 punts) Crea una funció anomenada Metricas_Riesgo_Sector que rebi el nom d'una columna de risc i torni un diccionari amb el valor màxim, el mínim i la mitjana.

1.4: (0,5 punts) Crea una funció anomenada Analisis_Adopcion_Període que rebi dos anys i torni un missatge formatat amb l'increment d'adopció d'IA.

```python
import pandas as pd

# 2.0. Carga de datos
df = pd.read_csv("data/usoDeInteligenciaArtificial.csv")

# 2.1: Función para filtrar ingresos
def Filtrar_Ingresos_Altos(umbral: float):
    """
    Filtra el DF por ingresos de software > umbral y ordena descendente.
    """
    filtro = df[df["AI Software Revenue(in Billions)"] > umbral]
    resultado = filtro.sort_values(by="AI Software Revenue(in Billions)", ascending=False)
    return resultado

print("\n 2.1: Años con ingresos > 50 Billions (ordenado)")
print(Filtrar_Ingresos_Altos(50)[["Year", "AI Software Revenue(in Billions)"]])

# 2.2: Función para volumen de empleo
def Calcular_Volumen_Empleo(data: pd.DataFrame):
    """
    Crea la columna Total_Jobs_Volume y retorna el DF.
    """
    # Creamos una copia para no modificar el original fuera de la función si no se desea
    df_mod = data.copy()
    df_mod["Total_Jobs_Volume"] = (
        df_mod["Estimated Jobs Eliminated by AI (millions)"] + 
        df_mod["Estimated New Jobs Created by AI (millions)"]
    )
    return df_mod

print("\n2.2: Primeras filas con la nueva columna de volumen de empleo")
df = Calcular_Volumen_Empleo(df) # Actualizamos nuestro df global
print(df[["Year", "Total_Jobs_Volume"]].head())

# 2.3: Función de métricas de riesgo por sector
def Metricas_Riesgo_Sector(columna_sector: str):
    """
    Calcula max, min y media de una columna de riesgo (limpiando el %).
    """
    # Convertir a numérico eliminando el símbolo % si está presente
    serie_numerica = df[columna_sector].replace('%', '', regex=True).astype(float)
    
    return {
        "Sector": columna_sector,
        "Maximo": serie_numerica.max(),
        "Minimo": serie_numerica.min(),
        "Media": round(serie_numerica.mean(), 2)
    }

print("\n 2.3: Análisis de riesgo en 'Manufacturing'")
print(Metricas_Riesgo_Sector("Jobs at High Risk of Automation - Manufacturing"))

# 2.4: Función de incremento de adopción en un periodo
def Analisis_Adopcion_Periodo(año_inicio: int, año_fin: int) -> str:
    """
    Calcula la variación de adopción entre dos años.
    """
    # Limpieza local para el cálculo
    adopcion_num = df["AI Adoption (%)"].str.replace('%', '').astype(float)
    
    # Obtener valores buscando por el índice del año
    val_inicio = adopcion_num[df["Year"] == año_inicio].values[0]
    val_fin = adopcion_num[df["Year"] == año_fin].values[0]
    
    incremento = val_fin - val_inicio
    
    return (f"Para el periodo {año_inicio}-{año_fin}, la adopción aumentó en "
            f"{incremento:.1f} puntos porcentuales.")

print("\n 2.4: Resultado del periodo 2018-2023")
print(Analisis_Adopcion_Periodo(2018, 2023))
```

3.1: (1 punt) Crea una funció que s'anomeni `job_impact_by_year` que rebi com a dades d'entrada:

- La ruta de l'arxiu
- Un any específic.

La funció tornarà un diccionari on la clau sigui l'any i el valor sigui una llista de tuples amb l'impacte laboral: {any : [('categoria', 'xx%'), ('categoria', 'xx%')]}

3.2: (1 punt) Crea una altra funció anomenada `automation_risk_dict` que rebi com a dades d'entrada:

- La ruta de l'arxiu
- Un llindar de risc (numèric).

La funció tornarà un diccionari on la clau és l'any i el valor és una llista de tuples amb els sectors que superen aquest percentatge de risc i el seu valor (Sector, Percentatge): {any : [('sector', 'xx%'), ('sector', 'xx%')]}

```python
#3.1

def job_impact_by_year(file_path: str, year: int) -> dict:
    df = pd.read_csv(file_path)
    # Convertimos a lista de tuplas (Year, Eliminated, Created)
    data = df[["Year", "Estimated Jobs Eliminated by AI (millions)", "Estimated New Jobs Created by AI (millions)"]].values.tolist()
    
    # Filtrado funcional por año
    filtered = list(filter(lambda x: x[0] == year, data))
    
    # Construcción del diccionario
    result = {}
    for y, elim, creat in filtered:
        result[y] = [("Eliminated", elim), ("Created", creat)]
    
    return result
```

```python
#test para comprobar el ejercicio 3.1
job_impact_by_year("data/usoDeInteligenciaArtificial.csv", 2020)
```

```python
#3.2

def automation_risk_dict(file_path: str, threshold: float) -> dict:
    df = pd.read_csv(file_path)
    # Limpiamos los nombres de columnas para facilitar el manejo
    cols = ["Year", "Jobs at High Risk of Automation - Transportation & Storage (%)", 
            "Jobs at High Risk of Automation - Wholesale & Retail Trade", 
            "Jobs at High Risk of Automation - Manufacturing"]
    
    data = df[cols].values.tolist()
    sectors = ["Transportation", "Retail", "Manufacturing"]
    
    resultado = {}
    for row in data:
        year = int(row[0])
        # Usamos list comprehension y filter para los sectores que superan el umbral
        high_risk = [(sectors[i-1], row[i]) for i in range(1, 4) if float(str(row[i]).replace('%','')) > threshold]
        
        if high_risk:
            resultado[year] = sorted(high_risk, key=lambda x: x[1], reverse=True)
            
    return resultado
```

# Exercici 4: Ús de funcions i expressions regulars

<span style="font-family: Courier New; background-color: #ffcc5c; color: #000000; padding: 3px;">EG</span> **(2 punts)**

4.1 (1 punt) Crea una funció anomenada `filter_automation_columns` que retorni un diccionari. La clau ha de ser el nom de la columna que contingui la paraula "Risk" seguida d'un guió o un espai, i el valor ha de ser la dada corresponent a l'any 2021.
Nota important. Les expressions regulars han de cercar paraules que continguin la paraula "risk" independentment que siguin majúscules o minúscules i ha de ser seguida d'un caràcter de puntuació o un espai en blanc.

4.2 (1 punt) Crea una funció anomenada `get_adoption_metrics` que identifiqui totes les columnes que parlin de "Adoption" o "Implement". La funció ha de tornar un diccionari on la clau sigui el nom de la columna i el valor sigui la dada del darrer any registrat al fitxer. Nota important: Les expressions regulars han de cercar paraules que continguin paraules amb les arrels "Adopt" o "Implement", sent vàlides qualsevol de les dues.

```python
#4.1
import re

def filter_automation_columns(file_path: str) -> dict:
    df = pd.read_csv(file_path)
    
    # Filtramos la fila del año 2021
    df_2021 = df[df['Year'] == 2021]
    
    # Patrón: Contiene "risk", seguido de espacio o guion
    pattern = re.compile(r"risk[\s\-]", re.IGNORECASE)
    
    # Buscamos qué columnas cumplen el patrón
    target_columns = [col for col in df.columns if pattern.search(col)]
    
    # Creamos el diccionario con los valores de esas columnas para el 2021
    # .iloc[0] porque sabemos que solo hay una fila para ese año
    result = {col: df_2021[col].iloc[0] for col in target_columns}
    
    return result

```

```python
#test para comprobar el ejercicio 4.1
filter_automation_columns("data/usoDeInteligenciaArtificial.csv")
```

```python
#4.2

def get_adoption_metrics(file_path: str) -> dict:
    df = pd.read_csv(file_path)
    
    # Patrón para encontrar "Adopt..." o "Implement..."
    # Usamos | para "o"
    pattern = re.compile(r"Adopt|Implement", re.IGNORECASE)
    
    # Seleccionamos columnas
    matched_cols = [c for c in df.columns if pattern.search(c)]
    
    # Obtenemos la última fila (el año más reciente)
    last_row = df.iloc[-1]
    
    # Diccionario resultante
    return {col: last_row[col] for col in matched_cols}

```

# Exercici 5: Ús de funcions i lectura de fitxers

<span style="font-family: Courier New; background-color: #ffcc5c; color: #000000; padding: 3px; ">EG</span> **(2.0 punts)**

Per a aquest exercici, es vol saber quines paraules són les que es repeteixen més d'una vegada en un text,
per tant, s'haurà de crear una funció que llegeixi un text i que calculi quantes vegades es repeteixen.
Finalment, es mostraran les paraules que es repeteixin més d'una vegada. El fitxer que cal utilitzar és: Elladron.docx.

Es demana crear la funció `lectura` que rebrà la ruta on es troba el fitxer i haurà de realitzar el següent:

1. Lectura del fitxer
2. Compteig de totes les paraules del fitxer
3. Filtrar les paraules que es repeteixin més d'una vegada.

El resultat daquesta funció seran les paraules (que es repeteixen més duna vegada) amb les vegades que es repeteix.

**Notes importants:**

1) Les paraules es compten per si soles, és a dir, si una paraula forma part d'una altra no compta.

2) Si una paraula té alguna lletra majúscula i una altra no, però és la mateixa paraula, es compten com a iguals, per tant, suma al comptador de la paraula.

Exemple:

<pre>
Frase: El menjar de la selva és millor que la lasanya.

Compteig de paraules:

la: 3
menjar: 1
de: 1
selva: 1
és: 1
millor: 1
que: 1
lasanya: 1

Resultat de la funció:

la: 3
</pre>

**Nota:**
Com que es tracta d'un fitxer .docx ha d'utilitzar la llibreria adequada per a la lectura.

```python
from docx import Document
from collections import Counter
import re

def lectura(MyDrive):
    # Leer el archivo Word
    doc = Document('data/Elladron.docx')

    # Extraer todo el texto del documento
    texto_completo = ""
    for parrafo in doc.paragraphs:
        texto_completo += parrafo.text + " "

    # Convertir todo el texto a minúsculas y dividir en palabras
    texto_completo = texto_completo.lower()

    # Usar una expresión regular para solo extraer palabras (ignorar números y puntuación)
    palabras = re.findall(r'\b\w+\b', texto_completo)

    # Contar la frecuencia de las palabras
    conteo_palabras = Counter(palabras)

    # Filtrar las palabras que se repiten más de una vez
    palabras_repetidas = {palabra: frecuencia for palabra, frecuencia in conteo_palabras.items() if frecuencia > 1}

    return palabras_repetidas

# Ejemplo de uso
ruta_docx = 'data/Elladron.docx'  # Ruta del archivo Word
palabras_repetidas = lectura(ruta_docx)

# Mostrar las palabras que se repiten más de una vez
print("Palabras que se repiten más de una vez:")
for palabra, frecuencia in palabras_repetidas.items():
    print(f"{palabra}: {frecuencia}")
```
